```python
def calculator(op, n1, n2): 
    if op == 'div': 
        return n1 / n2
    elif op == 'mul': 
        return n1 * n2 
```

```python
print (calculator('div', 4, 0))
print ("done")
```

```python
x = 0
try: 
    print ("Div 4 by 0 is:", calculator('div', 4, x))
except ZeroDivisionError: 
    print ("Do not ask me to divide by 0 please.")
    
print ("Done") 
```
