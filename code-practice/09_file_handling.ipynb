## CSV Files

```python
directory = "practice" 
import os 
csv_filename = os.path.join(directory, 'data_file.csv')
print (csv_filename )
```

```python
with open(csv_filename, 'r') as f: 
    for line in f: 
        print (line.strip().split(',') )
```

```python
rows = [] 

with open(csv_filename) as f: 
    for line in f: 
        rows.append(line.strip().split(','))
rows
```

```python
for r in rows[1:]:       # skip the header 
    for c in range(2, 6): 
        r[c] = float(r[c])
rows
```

```python
','.join( ['x', 'y', 'z'] )
```

```python
for r in rows: 
    print ([str(c) for c in r])
```

```python
out_rows = []

for r in rows: 
    r = [str(c) for c in r]    
    out_rows.append(r) 
```

```python
out_rows
```

```python
for r in out_rows: 
    out_line = ','.join(r)  # and then save this to a file 
    print (out_line)
```
