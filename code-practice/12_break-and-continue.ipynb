## Break

```python
for i in range(1, 6):
    if i == 3:
        break  
    print(i)

print("Loop dead...")
```

```python
for x in range(1, 11):
    square = x * x
    if square > 20:
        print("Square (",square,") is greater than 20! Stopping loop.")
        break  
        
    print("Number:" , x, "Square:" ,square)
```

### Continue

```python
for i in range(1, 6):
    if i == 3:
        continue  # Number 3 ko skip kar ke agle step par chala jayega
    print(i)

print("Loop is dead")
```

```python
for x in range(1, 11):
    if x % 2 != 0:
        continue  
    
    print("Even number:", x)
```
