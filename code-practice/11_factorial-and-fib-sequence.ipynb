### Factorial

```python
def factorial(num):
    fac = 1
    i = 1
    while i <= num:
        fac = i * fac
        i = i + 1
    return fac
```

```python
factorial(4)
```

### Fibonanci sequence
 0, 1, 1, 2, 3, 5, 8, 13, 21,...

```python
def fib(n): 
    if n == 0 or n == 1 :
        return n 
    
    else: 
        return fib(n-1) + fib(n-2)
```

```python
fib(4)
```

```python
fib(20)
```

```python
%time fib(20)
```

```python
%time fib(35)
```

```python
def fibnew(n): 
    if n <= 1:
        return n 
 
    a, b = 0, 1 
    
    for i in range(1, n + 1):
        n = a + b
        a = b
        b = n
        
    return a
```

```python
fibnew(4)
```

```python
%time fibnew(40)
```
