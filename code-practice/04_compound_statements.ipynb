```python
def check_number(x):
    if x < 0:
        print("number is negative")
    elif x > 0:
        print ("number is positive")
    else:
        print ("num is zero")
```

```python
check_number(3)
```

```python
def make_tea(sugar_spoon):
    if sugar_spoon == 0:
        print("sugar free tea is ready")
    elif sugar_spoon == 1:
        print(" tea is ready")
    else:
        print("sweet tea is ready=)")
```

```python
make_tea(2)
```
