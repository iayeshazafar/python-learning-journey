## Nested Sequence/list

```python
data = [2, 4, 6, 8, 10, 12]
pair= [
    [1, 10],
    [2, 20],
    [3, 30]    
]
```

```python
pair[2][1]
```

```python
for x, y in pair:
    print ("x =",x , "y=",y)
```

```python
pair = [[7, 7], [26, 24], [11,22], [26, 24], [6, 6], [1 + 1, 3]]
```

```python
same_count = 0      

for x, y in pair: 
    if x == y : 
        same_count += 1
    
print (same_count)
```
