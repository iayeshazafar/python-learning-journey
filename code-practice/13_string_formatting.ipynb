string formatting remove spaces like:

```python
print ("x=%d and y=%d" % (x, y))
```

```python
# alternate
print ("x={} and y={}".format(x, y))
```

```python
x = 2.637
y = 67
```

```python
print ("x=%.2f and y=%d" % (x, y))
```

### pading

```python
l = [5, 12, 26, 24, 134]
for i in l: 
    print ("%d" % i)
```

```python
l = [5, 12, 26, 24, 134]
for i in l: 
    print ("%4d" % i)
```

```python
for i in l: 
    print ("{:03}".format(i))
```
