### For loop
use when we know how many times a loop runs

```python
print ('Put sugar in.')
print ('Put milk in.')
print ('Put tea leaves in.')
```

```python
for ingredient in 'sugar', 'milk', 'tea leaves': 
    print ("Put", ingredient, 'in.')
    
```

```python
for i in range(1, 5+1):
    print (i,'Talha')
```

```python
for i in range (1, 10):
    if i % 2 == 0:
        print (i, "even no")
    else:
        print (i,'odd no')
```

## While loop
used when dont have idea how many times a loop run but know when to stop

```python
n = 0.13
i = 0

while n < 1:
    i = i + 1
    print (n)
    n = n + 0.13
    
print ('loop run', i, 'times')

    
```

```python
def save_money(target):
    savings = 0
    day = 1
    while savings > target:
        savings = savings + 100  
        print("Day:",day ,"Total Savings = Rs. ",savings)
        day = day + 1
```

```python
save_money(1000)
```
